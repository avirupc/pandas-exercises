### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
```

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv(url, sep="\t")
```

### Step 4. How many products cost more than $10.00?

```python
chipo["item_price"] = chipo["item_price"].str.replace("$", "", regex=False).astype(float)
chipo["product_price"] = chipo["item_price"] / chipo["quantity"]
```

```python
chipo[chipo["product_price"] > 10].drop_duplicates(["item_name", "choice_description"]).shape[0]
```

```python
# Alternate way: Create a new column with product plus choice variant
chipo["product_with_choice"] = chipo["item_name"] + " : " + chipo["choice_description"]
chipo[chipo["product_price"] > 10]["product_with_choice"].nunique()
```

### Step 5. What is the price of each item? 
###### print a data frame with only three columns item_name choice_description and product_price

```python
chipo[["item_name", "choice_description", "product_price"]].drop_duplicates().sort_values(by=["item_name", "product_price"])

# I think the author's solution is wrong. It operates on the filtered table with price > $10
```

### Step 6. Sort by the name of the item

```python
chipo.item_name.sort_values()

# OR

chipo.sort_values(by="item_name")
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
chipo.loc[chipo["item_price"].idxmax()]["quantity"]
```

```python
# Alternate, more robust
chipo.loc[chipo.item_price == chipo.item_price.max()]["quantity"]
```

Note: The solution provided by the author uses `idxmax` which does not give multiple rows in case of ties. The solution I mentioned as **Alternate** seems to be more robust. <br> For this particular dataset and query, this does not matter, but readers can do a quick check by comparing results using `idxmin` and `chipo.item_price == chipo.item_price.min()` and see the difference 🙂

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
chipo[chipo["item_name"] == "Veggie Salad Bowl"]["quantity"].sum()
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
chipo[(chipo.item_name == "Canned Soda") & (chipo.quantity > 1)].shape[0]
```
